```python
def main():
    # Initialize data for 10 salesmen and 5 items
    salesmen = []
    sales_data = []
    
    print("Enter data for 10 salesmen:")
    
    # Input data for each salesman
    for i in range(10):
        print(f"\nSalesman {i+1}:")
        name = input("Name: ")
        print(f"Enter sales for {name}:")
        
        # Input sales for 5 items
        sales = []
        for item in range(1, 6):
            sales.append(float(input(f"Item{item}: ")))
        
        salesmen.append(name)
        sales_data.append(sales)
    
    # Calculate totals
    totals = [sum(sales) for sales in sales_data]
    grand_total = sum(totals)
    
    # Print the report (exactly matching the required format)
    print("\n| Name    | Item1 | Item2 | Item3 | Item4 | Item5 | TotalSales |")
    print("-" * 65)
    
    for i in range(10):
        print(f"| {salesmen[i]:<8}|", end="")
        for sale in sales_data[i]:
            print(f" {sale:<6.0f}|", end="")
        print(f" {totals[i]:<10.0f}|")
    
    print(f"\nGrandTotal {grand_total:.0f}")

if __name__ == "__main__":
    main()
```
