```python
# Number 2
def convert(fahrenheit):
    celcius = 5/9 *(fahrenheit-32)
    
    return celcius

fahrenheit = int(input("Enter the temperature in fahrenheit"))    
celcius = convert(fahrenheit)   
print(f"It is {celcius} degrees Celcius")

if celcius > 20:
    print("IT'S HOT HERE")
else:
    print("IT'S COLD HERE")
```
