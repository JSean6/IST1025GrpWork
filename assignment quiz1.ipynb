```python
import random

def play_game():
    print("Welcome to Stone-Knife-Paper Game!")
    print("Enter your choice:")
    print("S for Stone")
    print("K for Knife")
    print("P for Paper")
    
    # Computer's random choice (0=S, 1=K, else P)
    computer_num = random.randint(0, 2)
    if computer_num == 0:
        computer_choice = 'S'
    elif computer_num == 1:
        computer_choice = 'K'
    else:
        computer_choice = 'P'
    
    
    while True:
        player_choice = input("Your choice (S/K/P): ").upper()
        if player_choice in ['S', 'K', 'P']:
            break
        print("Invalid input. Please enter S, K, or P.")
    
    print(f"\nYour choice: {player_choice}")
    print(f"Computer's choice: {computer_choice}")
    

    if player_choice == computer_choice:
        print("It's a tie!")
    elif (
        (player_choice == 'S' and computer_choice == 'K') or
        (player_choice == 'K' and computer_choice == 'P') or
        (player_choice == 'P' and computer_choice == 'S')
    ):
        print("You win!")
    else:
        print("Computer wins!")


play_game()
```
